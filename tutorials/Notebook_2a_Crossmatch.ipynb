# Cross-matching DESI Legacy Survey with Rubin DP2 photo-z

In this notebook we'll show how to:
- Load the DESI Legacy Survey DR11 South catalog
- Cross-match it with the Rubin DP2 photo-z catalog
- Flag objects where the Legacy Survey and Rubin photo-z agree
- Save the flags as a HATS catalog, which can be shared with non-DRHs

> [!NOTE]
> This notebook uses Rubin DP2 data, so it can be run by Rubin data rights holders (DRHs) only, e.g. on the [Rubin Science Platform](https://data.lsst.cloud) with the latest weekly image.
> Since we would like to share the notebook and its outputs with everyone, we never print or plot individual rows of the DP2 catalog: all the outputs are either catalog metadata or aggregated statistics.

````python
# lsdb is pre-installed on RSP, uncomment the line below if you are on another platform
# %pip install -qq lsdb
````

````python
# Standard library imports
import logging
import os
import warnings

# Third-party imports
import astropy.units as u
import dask
import lsdb
import matplotlib.pyplot as plt
import numpy as np
from astropy.coordinates import SkyCoord
from astropy.visualization.wcsaxes import Quadrangle
from dask.distributed import Client
from lsdb import BoxSearch

# Configuration
warnings.simplefilter("ignore")
logging.getLogger("distributed").setLevel(logging.WARNING)
dask.config.set({"dataframe.convert-string": False})
````

## Setup Dask cluster

We use the same Dask setup as in `Notebook_1a_Intro`: sixteen workers, one per CPU of a "Large" RSP container, with one thread and 2 GB of memory each, which is enough for this low-memory pipeline, and `/deleted-sundays/` for spilling data to disk.

````python
client = Client(
    n_workers=16,
    threads_per_worker=1,
    local_directory=f"/deleted-sundays/{os.environ.get('USER', 'dask_scratch')}",
)

# Print the dashboard link and port
print(f"Dask is running at: {client.dashboard_link}")
````

## Select a region on the sky

We select a ~10×10 degree box, which should be covered by both DP2 and the DESI Legacy Survey. `BoxSearch` selects objects within the given ranges of right ascension and declination. The box is not far from the Galactic plane, so we expect the interstellar extinction to vary across the box.

````python
# Right ascension and declination ranges, in degrees
RA_RANGE = (295.0, 305.0)
DEC_RANGE = (-25.0, -15.0)

box_search = BoxSearch(ra=RA_RANGE, dec=DEC_RANGE)


def plot_partitions(catalog, title):
    """Plot the HATS partitions of a catalog, zoomed into our box, and the box in red."""
    fig, ax = catalog.plot_pixels(
        projection="AIT",
        center=SkyCoord(ra=300.0, dec=-20.0, unit="deg"),
        fov=(14 * u.deg, 14 * u.deg),
        plot_title=title,
    )
    # Quadrangle follows the lines of constant RA and Dec, as BoxSearch does
    box = Quadrangle(
        (RA_RANGE[0], DEC_RANGE[0]) * u.deg,
        (RA_RANGE[1] - RA_RANGE[0]) * u.deg,
        (DEC_RANGE[1] - DEC_RANGE[0]) * u.deg,
        edgecolor="red",
        facecolor="none",
        lw=2,
        transform=ax.get_transform("icrs"),
    )
    ax.add_patch(box)
    plt.show()
````

## Open the DESI Legacy Survey DR11 South catalog

The [DESI Legacy Survey DR11 South](https://data.lsdb.io/DESI/Legacy_Survey_DR11.0_South) catalog is public and served in HATS format over HTTP, so we can open it from anywhere. We need only the object ID and the photo-z point estimate, see the [column descriptions](https://www.legacysurvey.org/dr11/files/#sweep-catalogs-region-sweep) for other columns.

````python
LEGACY_SURVEY_URL = "http://slac.data.lsdb.io:10942/lsdb/legacysurvey_dr11.0_south"

legacy_survey = lsdb.open_catalog(
    LEGACY_SURVEY_URL,
    # coordinates and _healpix_29 are always loaded
    columns=["LS_ID_DR11", "Z_PHOT_MEDIAN"],
    search_filter=box_search,
)
# Keep objects with a valid photo-z
legacy_survey = legacy_survey.query("Z_PHOT_MEDIAN >= 0")
legacy_survey
````

The Legacy Survey data are public, so we can freely look at them:

````python
legacy_survey.head()
````

## Open the DP2 photo-z catalog

The DP2 photo-z catalog has photo-z estimates from several algorithms. For each algorithm there are four point estimates, `{algorithm}_z_best`, `_z_mean`, `_z_median`, and `_z_mode`, and credible intervals. We open the catalog with its default columns, which include the `_z_best` estimate of each algorithm, and use them to list the algorithms. For the flag below we use `fzboost_z_best`.

Unlike the Legacy Survey, these are Rubin data, so we don't look at the rows, only at the catalog structure.

````python
# DP2 photo-z on RSP, check https://lsdb.io/dp2 for the paths on other platforms
dp2_photoz = lsdb.open_catalog(
    "/rubin/lsdb_data/dp2/object_photoz",
    search_filter=box_search,
)
dp2_photoz
````

````python
[column.removesuffix("_z_best") for column in dp2_photoz.columns if column.endswith("_z_best")]
````

## Cross-match

`crossmatch()` finds, for each object of the left catalog, the nearest object of the right catalog within the given radius. LSDB cross-matches catalogs partition by partition: for each pair of overlapping HEALPix tiles it runs a separate task, so the cross-match is parallel and never loads the full catalogs into memory.



The order of the catalogs matters for us: the result inherits the `_healpix_29` index of the left catalog. We use the Legacy Survey as the left catalog, so `_healpix_29` of the result comes from the public Legacy Survey coordinates, not from the Rubin ones.

````python
xmatch = legacy_survey.crossmatch(
    dp2_photoz,
    radius_arcsec=1.0,
    suffix_method="overlapping_columns",
    suffixes=("_legacy_survey", "_dp2"),
)
xmatch
````

Let's look at the HATS partitions of both catalogs and of the cross-match, as we did in `Notebook_1a_Intro`, but zoomed into our box, which we draw in red. `.search()` keeps only the partitions overlapping the box, so the partitions on the edges stick out of it, but only the rows inside the box will be loaded. Partition maps show the catalog structure, not the data, so the DP2 map is safe to share.

The cross-match is lazy, but its partitions are already known: LSDB aligns the partitions of the two catalogs, keeping only the sky covered by both, and where the partitions have different sizes, the result uses the smaller ones.

````python
plot_partitions(legacy_survey, "DESI Legacy Survey DR11 South partitions")
plot_partitions(dp2_photoz, "Rubin DP2 photo-z partitions")
plot_partitions(xmatch, "Cross-match partitions")
````

## Flag photo-z agreement

We can't share the Rubin photo-z values with non-DRHs, but we can share a coarse derived product: a flag telling whether the Rubin photo-z is within 0.1 of the Legacy Survey one.

````python
def add_photoz_match_flag(df):
    """Add a flag telling whether DP2 and Legacy Survey photo-z agree within 0.1."""
    return df.assign(lsdr11_rubindp2_photoz_match=np.abs(df["fzboost_z_best"] - df["Z_PHOT_MEDIAN"]) < 0.1)


flagged = xmatch.map_partitions(add_photoz_match_flag)
````

## Select the columns to share

Before saving anything, we explicitly select the columns which are safe to share:
- `_healpix_29`, the catalog index: it comes from the Legacy Survey coordinates;
- `LS_ID_DR11`, `RA`, `DEC`, and `Z_PHOT_MEDIAN`: public Legacy Survey data;
- `objectId`: Rubin object IDs can be shared;
- `lsdr11_rubindp2_photoz_match`: our derived flag.

We drop the Rubin photo-z values (`*_z_best` columns), the Rubin coordinates (`ra` and `dec`), and the cross-match distance (`_dist_arcsec`), which is derived from the Rubin coordinates.

````python
shareable = flagged[["LS_ID_DR11", "RA", "DEC", "objectId", "Z_PHOT_MEDIAN", "lsdr11_rubindp2_photoz_match"]]
shareable
````

## Save as HATS catalog

`write_catalog()` runs the whole pipeline: it reads both catalogs, cross-matches them, computes the flags, and saves the result partition by partition.

````python
shareable.write_catalog("lsdr11_rubindp2_photoz_match", overwrite=True)
````

## What non-DRHs will see

The saved catalog contains only the columns we have selected, so now we can look at the rows.

````python
shared = lsdb.open_catalog("lsdr11_rubindp2_photoz_match")
shared.head()
````

````python
n_match = shared["lsdr11_rubindp2_photoz_match"].sum().compute()
print(f"Number of matched objects: {len(shared):,}")
print(f"Fraction with matching photo-z: {n_match / len(shared):.1%}")
````
